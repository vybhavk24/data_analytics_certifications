```python
%%sql
CREATE TABLE IF NOT EXISTS SalesDataset (
    Order_Id INTEGER PRIMARY KEY AUTOINCREMENT,
    Amount REAL,
    Profit REAL,
    Quantity INTEGER,
    Category TEXT,
    "Sub-category" TEXT,
    PaymentMode TEXT,
    Order_date TEXT,
    CustomerName TEXT,
    State TEXT,
    City TEXT,
    "Year-Month" TEXT
);
```

```python
%%sql
SELECT * FROM SalesDataset LIMIT 5;
```
