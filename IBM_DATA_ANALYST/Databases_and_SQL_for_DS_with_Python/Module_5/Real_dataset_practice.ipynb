```python
import pandas as pd

# Load the CSV into a DataFrame
df = pd.read_csv('sales_dataset_practice.csv')
```

```python
print("Initial data preview:")
print(df.head())
```

```python
# Check missing values
print("Missing values per column:")
print(df.isnull().sum())

# Fill missing numeric values in 'Amount' and 'Profit'
df['Amount'] = df['Amount'].fillna(df['Amount'].mean())
df['Profit'] = df['Profit'].fillna(df['Profit'].mean())

# Fill missing categorical values in 'State'
df['State'] = df['State'].fillna('Unknown')

# Optional: Remove duplicate rows based on Order_Id (if Order_Id is unique)
df = df.drop_duplicates(subset='Order_Id')

print("After cleaning missing values:")
print(df.head())
```

```python
# Convert Order_date to datetime, allowing for multiple formats
df['Order_date'] = pd.to_datetime(df['Order_date'], errors='coerce')

# In case some dates still fail, they become NaT (Not a Time); you might choose to drop or fill them.
# Let's drop rows with invalid Order_date:
df = df.dropna(subset=['Order_date'])

print("Data types after conversion:")
print(df.dtypes)
```

```python
print("Descriptive statistics:")
print(df.describe())
print("\nDataframe Info:")
print(df.info())
```

```python
# Group by Category to see total sales and average profit per category
category_summary = df.groupby('Category').agg(
    TotalSales=('Amount', 'sum'),
    AverageProfit=('Profit', 'mean'),
    TotalOrders=('Order_Id', 'count')
)
print("\nSales summary by Category:")
print(category_summary)
```

```python
# Create a new column for Profit Margin
df['ProfitMargin'] = df['Profit'] / df['Amount']
print("\nData with ProfitMargin:")
print(df[['Order_Id', 'Amount', 'Profit', 'ProfitMargin']].head())
```
