### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here

# Display column data types and non-null counts
print("Dataset Info:\n")
print(df.info())

# Display missing values per column
print("\nMissing Values Per Column:\n")
print(df.isnull().sum())
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here

print("Basic Statistics for Numerical Columns:\n")
print(df.describe())
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here

# Display unique country values
unique_countries = df['Country'].unique()
print("Unique values in 'Country':\n")
print(unique_countries)

# Display value counts for each country (to spot irregular entries)
print("\nCountry Value Counts:\n")
print(df['Country'].value_counts(dropna=False))
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here

country_mapping = {
    'United States of America': 'United States',
    'USA': 'United States',
    'UK': 'United Kingdom',
    'U.K.': 'United Kingdom',
    'England': 'United Kingdom',
    'Republic of India': 'India',
    'Viet Nam': 'Vietnam',
}

df['Country'] = df['Country'].replace(country_mapping)

# Example: Standardizing 'EdLevel' values
# Optional cleanup: convert values to lowercase and strip extra spaces for uniformity
df['EdLevel'] = df['EdLevel'].str.strip().str.lower()

# Mapping example if needed
edlevel_mapping = {
    'bachelor’s degree': 'Bachelor’s degree',
    'master’s degree': 'Master’s degree',
    'some college/university study without earning a degree': 'Some college',
    'secondary school (e.g. american high school, german realschule or hauptschule, etc.)': 'Secondary school',
    # Add more mappings if needed
}

df['EdLevel'] = df['EdLevel'].replace(edlevel_mapping)

# Verify standardization
print("Standardized 'Country' values:\n", df['Country'].value_counts().head())
print("\nStandardized 'EdLevel' values:\n", df['EdLevel'].value_counts().head())
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here

employment_dummies = pd.get_dummies(df['Employment'], prefix='Employment')

# Concatenate the new one-hot encoded columns with the original dataframe
df_encoded = pd.concat([df, employment_dummies], axis=1)

# Optional: drop the original 'Employment' column if no longer needed
# df_encoded = df_encoded.drop('Employment', axis=1)

# Display the first few rows of the updated dataframe
print(df_encoded.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here

missing_counts = df.isnull().sum().sort_values(ascending=False)
print("Columns with the highest number of missing values:\n")
print(missing_counts.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here

# Check for missing values first
missing_before = df['ConvertedCompYearly'].isnull().sum()
print(f"Missing values in 'ConvertedCompYearly' before imputation: {missing_before}")

# Impute with median
median_value = df['ConvertedCompYearly'].median()
df['ConvertedCompYearly'].fillna(median_value, inplace=True)

# Verify imputation
missing_after = df['ConvertedCompYearly'].isnull().sum()
print(f"Missing values in 'ConvertedCompYearly' after imputation: {missing_after}")
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here

# Check for missing values first
missing_before = df['RemoteWork'].isnull().sum()
print(f"Missing values in 'RemoteWork' before imputation: {missing_before}")

# Find the most frequent value (mode)
most_frequent_value = df['RemoteWork'].mode()[0]

# Impute missing values with the mode
df['RemoteWork'].fillna(most_frequent_value, inplace=True)

# Verify imputation
missing_after = df['RemoteWork'].isnull().sum()
print(f"Missing values in 'RemoteWork' after imputation: {missing_after}")
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here

# Ensure no missing values before scaling
df = df[df['ConvertedCompYearly'].notnull()]

# Calculate min and max
min_val = df['ConvertedCompYearly'].min()
max_val = df['ConvertedCompYearly'].max()

# Apply Min-Max scaling
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_val) / (max_val - min_val)

# Display first few rows to verify
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_MinMax']].head())
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np
# Ensure there are no missing or zero values (log(0) is undefined)
df = df[df['ConvertedCompYearly'] > 0]

# Apply log transformation
df['ConvertedCompYearly_Log'] = np.log(df['ConvertedCompYearly'])

# Display first few rows to verify
print(df[['ConvertedCompYearly', 'ConvertedCompYearly_Log']].head())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here

# First, replace non-numeric values with approximate numbers
df['YearsCodePro'] = df['YearsCodePro'].replace({
    'Less than 1 year': 0.5,
    'More than 50 years': 51
})

# Convert the column to numeric
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Define function to assign experience levels
def categorize_experience(years):
    if pd.isnull(years):
        return 'Unknown'
    elif years < 2:
        return 'Beginner'
    elif 2 <= years < 5:
        return 'Intermediate'
    elif 5 <= years < 10:
        return 'Advanced'
    else:
        return 'Expert'

# Apply the function to create the new column
df['ExperienceLevel'] = df['YearsCodePro'].apply(categorize_experience)

# Display the value counts for the new column
print(df['ExperienceLevel'].value_counts())
```
