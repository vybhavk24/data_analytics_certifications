### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here

!pip install seaborn
import seaborn as sns

# Filter relevant column and drop missing values
df_db_pie = df[['DatabaseWantToWorkWith']].dropna()

# Split multiple databases and explode into separate rows
df_db_pie['DatabaseWantToWorkWith'] = df_db_pie['DatabaseWantToWorkWith'].str.split(';')
df_db_pie = df_db_pie.explode('DatabaseWantToWorkWith')

# Count occurrences of each database
db_counts = df_db_pie['DatabaseWantToWorkWith'].value_counts().head(5)

# Plot the pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    db_counts,
    labels=db_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel')
)

plt.title('Top 5 Databases Respondents Want to Work With', fontsize=14)
plt.tight_layout()
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here

# Filter the relevant column and drop missing values
df_devtype_pie = df[['DevType']].dropna()

# Split multiple developer types and explode into individual rows
df_devtype_pie['DevType'] = df_devtype_pie['DevType'].str.split(';')
df_devtype_pie = df_devtype_pie.explode('DevType')

# Count the occurrences of each developer type
devtype_counts = df_devtype_pie['DevType'].value_counts().head(5)

# Plot the pie chart
plt.figure(figsize=(8, 8))
plt.pie(
    devtype_counts,
    labels=devtype_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('pastel')
)

plt.title('Top 5 Developer Roles Among Respondents', fontsize=14)
plt.tight_layout()
plt.show()
```
